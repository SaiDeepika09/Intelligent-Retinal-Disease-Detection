```python
import numpy as np

# Inputs
X = np.array([2.0, 3.0])

# Weights
W = np.array([0.5, 0.2])

# Bias
b = 0.1

# Weighted sum
z = np.dot(X, W) + b

print("Inputs:", X)
print("Weights:", W)
print("Bias:", b)
print("Weighted sum:", z)
```

```python
# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))


# Tanh function
def tanh(x):
    return np.tanh(x)


# ReLU function
def relu(x):
    return np.maximum(0, x)


print("Input (z):", z)
print("Sigmoid:", sigmoid(z))
print("Tanh:", tanh(z))
print("ReLU:", relu(z))
```

```python
# Activation function
def relu(x):
    return np.maximum(0, x)


# Forward propagation function
def forward_propagation(X, W, b):

    # Step 1: Calculate weighted sum
    z = np.dot(X, W) + b

    # Step 2: Apply activation function
    output = relu(z)

    return z, output


# Inputs
X = np.array([2.0, 3.0])

# Weights
W = np.array([0.5, 0.2])

# Bias
b = 0.1


# Perform forward propagation
z, output = forward_propagation(X, W, b)

print("Weighted Sum (z):", z)
print("Output after ReLU:", output)
```

```python
# Actual/expected value
y_true = 2.0

# Model prediction
y_pred = output

# Calculate Mean Squared Error
loss = (y_true - y_pred) ** 2

print("Actual value:", y_true)
print("Predicted value:", y_pred)
print("Loss (MSE):", loss)
```

Back Propagation

```python
# Learning rate
learning_rate = 0.01

# Forward propagation values
z = np.dot(X, W) + b
y_pred = relu(z)

# Calculate error
error = y_pred - y_true

# Gradient of loss with respect to prediction
gradient_output = 2 * error

# Since z > 0, ReLU derivative = 1
gradient_z = gradient_output * 1

# Gradients for weights
gradient_W = gradient_z * X

# Gradient for bias
gradient_b = gradient_z

# Update weights and bias
W = W - learning_rate * gradient_W
b = b - learning_rate * gradient_b

print("Updated weights:", W)
print("Updated bias:", b)
```

```python
# Forward propagation again using updated parameters

z_new = np.dot(X, W) + b
y_pred_new = relu(z_new)

# Calculate new loss
loss_new = (y_true - y_pred_new) ** 2

print("New prediction:", y_pred_new)
print("Actual value:", y_true)
print("New loss:", loss_new)
```

1. Input
      ↓
2. Forward Propagation
      ↓
3. Prediction
      ↓
4. Calculate Loss
      ↓
5. Backpropagation
      ↓
6. Calculate Gradients
      ↓
7. Update Weights and Bias
      ↓
8. Forward Propagation Again
      ↓
9. Better Prediction

```python
# Reset parameters so we start fresh
W = np.array([0.5, 0.2])
b = 0.1

# Settings
learning_rate = 0.01
epochs = 10

# Store loss values
loss_history = []

for epoch in range(epochs):

    # -------------------------
    # 1. Forward Propagation
    # -------------------------
    z = np.dot(X, W) + b
    y_pred = relu(z)

    # -------------------------
    # 2. Calculate Loss
    # -------------------------
    loss = (y_true - y_pred) ** 2
    loss_history.append(loss)

    # -------------------------
    # 3. Backpropagation
    # -------------------------
    error = y_pred - y_true

    gradient_output = 2 * error

    # ReLU derivative
    gradient_z = gradient_output if z > 0 else 0

    # Gradients
    gradient_W = gradient_z * X
    gradient_b = gradient_z

    # -------------------------
    # 4. Update Parameters
    # -------------------------
    W = W - learning_rate * gradient_W
    b = b - learning_rate * gradient_b

    print(
        f"Epoch {epoch + 1}: "
        f"Prediction = {y_pred:.4f}, "
        f"Loss = {loss:.6f}"
    )

print("\nFinal Weights:", W)
print("Final Bias:", b)
```

```python
# ReLU function
def relu(x):
    return np.maximum(0, x)


# Function to train the simple neuron
def train_model(learning_rate, epochs=10):

    # Reset parameters for each experiment
    W = np.array([0.5, 0.2])
    b = 0.1

    losses = []

    for epoch in range(epochs):

        # Forward propagation
        z = np.dot(X, W) + b
        y_pred = relu(z)

        # Calculate loss
        loss = (y_true - y_pred) ** 2
        losses.append(loss)

        # Backpropagation
        error = y_pred - y_true
        gradient_output = 2 * error

        # ReLU derivative
        gradient_z = gradient_output if z > 0 else 0

        # Gradients
        gradient_W = gradient_z * X
        gradient_b = gradient_z

        # Update parameters
        W = W - learning_rate * gradient_W
        b = b - learning_rate * gradient_b

    return y_pred, losses[-1], losses


# Different learning rates
learning_rates = [0.001, 0.01, 0.1]

for lr in learning_rates:

    prediction, final_loss, losses = train_model(lr)

    print(f"\nLearning Rate: {lr}")
    print(f"Final Prediction: {prediction:.4f}")
    print(f"Final Loss: {final_loss:.6f}")
```
